```python
import h5py
import numpy as np
import matplotlib.pyplot as plt


# Open file
f = h5py.File('desi_dr1_extracted_spectra.h5', 'r')


# Access first BGS spectrum
bgs = f['ELG/spectra/obj_0008']
wave = bgs['wavelength'][:]  # Angstroms
flux = bgs['flux'][:]        # 10^-17 erg/s/cm²/Å
ivar = bgs['ivar'][:]        # Inverse variance


# Plot
plt.figure(figsize=(12, 4))
plt.plot(wave[flux>0], flux[flux>0], 'k-', linewidth=0.5)
plt.xlabel('Wavelength (Å)')
plt.ylabel('Flux (10⁻¹⁷ erg/s/cm²/Å)')
plt.title(f'TARGETID: {bgs.attrs["targetid"]}')
plt.show()


# Access catalog data
catalog = f['BGS/catalog']
ra = catalog['ra'][:]        # degrees
dec = catalog['dec'][:]      # degrees
targetids = catalog['targetid'][:]
```
